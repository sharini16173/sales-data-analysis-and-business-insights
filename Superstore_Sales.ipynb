```python
#Data cleaning:-
df.columns = df.columns.str.strip()
print(df.columns)
```

```python
#Convert datatype str to date
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y', errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%m/%d/%Y', errors='coerce')
print(df[['Order Date','Ship Date']].dtypes)
```

```python
#Inspect invalid values
print(df[df["Sales"] <= 0])
print(df[df["Quantity"]<= 0])
```

```python
print(df[df["Profit"]<=0])
```

```python
#Create useful date column-[ Extract year and month ]
df['Year']=df['Order Date'].dt.year
df['Month']=df['Order Date'].dt.month
df['Month Name']=df['Order Date'].dt.month_name()
df['Year-Month']=df['Order Date'].dt.to_period('M').astype(str)
print(df[['Order Date','Year','Month','Month Name','Year-Month']].head())
```

```python
#Calculate shipping duration
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
print(df[["Order Date", "Ship Date", "Shipping Days"]].head())
```

```python
#Add profit status column
df["Profit Status"] = "Profit"
df.loc[df["Profit"] < 0, "Profit Status"] = "Loss"
print(df[["Profit", "Profit Status"]].head())
```

```python
df.to_csv("outputs/cleaned_sales.csv",index=False)
```

```python
#Calculate the main business KPIs
total_sales=df["Sales"].sum()
total_quantity=df["Quantity"].sum()
total_profit=df["Profit"].sum()
total_order=df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

print("Total_Sales:",round(total_sales,2))
print("Total_Quantity:",total_quantity)
print("Total_Profit:",round(total_profit,2))
print("Total_Orders:",total_order)
print("Total_Customers:", total_customers)
```

```python
#Calculate average order value and profit margin
if total_order >0:
    average_order_value = total_sales / total_order
    average_profit_per_order = total_profit / total_order
else:
    average_order_value = 0
    average_profit_per_order = 0
    
if total_sales >0:
     profit_margin = (total_profit / total_sales) * 100
else:
    profit_margin=0
print("Average Order Value:",round(average_order_value,2))
print("Average Profit per Order:",round(average_profit_per_order,2))
print("Profit Margin:",round(profit_margin,2),"%")
```

```python
# Customer Analysis
latest_year = df["Year"].max()
active_customers = df[df["Year"] == latest_year]["Customer ID"].nunique()
inactive_customers = total_customers - active_customers

print("Customers who ordered in", latest_year, ":", active_customers)
print("Customers who did not order in", latest_year, ":", inactive_customers)
```

```python
#Top 10 customers by sales
top_customers = df.groupby("Customer Name")["Sales"].sum()
print(top_customers.sort_values(ascending=False).head(10))
```

```python
#Compare product category
category_sales=df.groupby("Category")["Sales"].sum()
print(category_sales.sort_values(ascending=False))
```

```python
#Category with Highest sales
category_analysis=df.groupby("Category").agg(
    Total_Sales=("Sales","sum"),
    Total_Quantity=("Quantity","sum"),
    Total_Profit=("Profit","sum"))
print(category_analysis.sort_values("Total_Sales",ascending=False))                                                                                                                      
```

```python
#Compare region sales
region_sales=df.groupby("Region")["Sales"].sum()
print(region_sales.sort_values(ascending=False))
```

```python
#Region with Highest sales
region_analysis=df.groupby("Region").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum"))
print(region_analysis.sort_values("Total_Sales",ascending=False))
```

```python
#Compare Profitable subcategory
subcategory_profit=df.groupby("Sub-Category")["Profit"].sum()
print(subcategory_profit.sort_values(ascending=False))
```

```python
#5 Highest-profit subcategories
print(subcategory_profit.sort_values(ascending=False).head(5))
```

```python
#5 Lowest-profit subcategories
print(subcategory_profit.sort_values().head(5))
```

```python
#Finding best selling products
product_sales=df.groupby("Product Name")["Sales"].sum()
print(product_sales.sort_values(ascending=False).head(10))
```

```python
#Analyse Monthly sales-[sales for each month]
monthly_sales=df.groupby("Year-Month")["Sales"].sum()
print(monthly_sales)
```

```python
#Highest monthly sales
print(monthly_sales.sort_values(ascending=False).head())
```

```python
#Lowest monthly sales
print(monthly_sales.sort_values().head())
```

```python
#Compare month-on-month sales growth-[sales with the previous month]
monthly_sales = monthly_sales.sort_index()
monthly_growth = monthly_sales.pct_change() * 100
print(monthly_growth.round(2))
```

```python
#Monthly KPI comparison
monthly_kpi = df.groupby("Year-Month").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit","sum"),
    Orders=("Order ID", "nunique"),
    Customers=("Customer ID", "nunique"))
print(monthly_kpi.tail())
```

```python
# Month-on-Month KPI Change
monthly_kpi_change = monthly_kpi[["Orders", "Customers"]].pct_change() * 100
print(monthly_kpi_change.round(2))
```

```python
#Compare customer segments
segment_analysis = df.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"))
print(segment_analysis)
```

```python
#Analyse discounts and profit-[Check the total profit at each discount level]
discount_analysis=df.groupby("Discount").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum"),
    Total_Quantity=("Quantity","sum"))
print(discount_analysis.sort_index())
```

```python
#Calculate profit margin by category
category_profit=df.groupby("Category").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum"))
category_profit["Profit_Margin"]=(category_profit["Total_Profit"]/ category_profit["Total_Sales"] * 100)
print(category_profit.round(2))
```

```python
#Find orders with negative profit
order_analysis=df.groupby("Order ID").agg(
    Total_Sales=("Sales","sum"),
    Total_Profit=("Profit","sum")
)
loss_orders=order_analysis[order_analysis["Total_Profit"]<0]
print(loss_orders.head(10))
print("Number of loss-making orders:", len(loss_orders))
```

```python
#import the Chart libraries:--
import matplotlib.pyplot as plt
import os
import seaborn as sns
```

```python
#Monthly Sales trend-Line Chart
df["Year-Month"] = df["Order Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("Year-Month")["Sales"].sum()
monthly_sales = monthly_sales.sort_index()

monthly_sales.plot(kind='line', marker='o', figsize=(12,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Year-Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("outputs/monthly_sales.png")

plt.show()
```

```python
#Sales by category-Bar chart
category_sales = df.groupby("Category")["Sales"].sum()
category_sales = category_sales.sort_values(ascending=False)

category_sales.plot(kind="bar",edgecolor="black", figsize=(8, 5))

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("outputs/category_sales.png")
plt.show()
```

```python
#Profit by subcategory-Horizontal bar chart
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum()
subcategory_profit = subcategory_profit.sort_values()

subcategory_profit.plot(kind="barh",edgecolor="black", figsize=(9, 7))

plt.title("Profit by Subcategory")
plt.xlabel("Total Profit")
plt.ylabel("Subcategory")
plt.tight_layout()
plt.savefig("outputs/subcategory_profit.png")
plt.show()
```

```python
#Compare sales by region-Region chart
region_sales = df.groupby("Region")["Sales"].sum()
region_sales = region_sales.sort_values(ascending=False)

region_sales.plot(kind="bar",edgecolor="black", figsize=(8, 5))

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("outputs/region_sales.png")
plt.show()
```

```python
#View the distribution of sales-Histogram
df["Sales"].dropna().plot(
    kind="hist",
    bins=30,
    edgecolor="black",
    figsize=(8, 5)
)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Number of Records")
df[df["Sales"] <= 1000]["Sales"]

plt.tight_layout()
plt.savefig("outputs/sales_distribution.png")
plt.show()
```

```python
#Visualise discounts and profit-Scatter plot
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Discount", y="Profit", alpha=0.5)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.savefig("outputs/discount_profit.png")
plt.show()
```

```python
#Create a summary table of KPIs:-
kpi_summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Quantity",
        "Total Orders",
        "Average Order Value",
        "Profit Margin",
        "Total Customers"
    ],
    "Value": [
        total_sales,
        total_profit,
        total_quantity,
        total_order,
        average_order_value,
        profit_margin,
        total_customers
    ]
})
print(kpi_summary)
```

```python
#Export the analyis to Excel:-
insights_data = {
    "Category": [
        "Key Insight", 
        "Key Insight", 
        "Key Insight", 
        "Key Insight",
        "Strategic Recommendation", 
        "Strategic Recommendation", 
        "Strategic Recommendation"
    ],
    "Findings & Actions": [
        "Applying discounts of 30% or more consistently results in financial losses.",
        "Furniture category drives high sales volume but yields a critically low profit margin.",
        "Tables and Bookcases are actively loss-making sub-categories.",
        "Sales volume demonstrates strong seasonality, peaking during September, November, and December.",
        "Cap standard promotional discounts at a strict 20% ceiling to protect baseline profitability.",
        "Reevaluate supplier costs and pricing structures specifically for Tables and Bookcases.",
        "Direct Q4 promotional budgets toward categories with higher profit margins."
    ]
}

insights_df = pd.DataFrame(insights_data)

# Export the entire analysis including the new Insights sheet to Excel
with pd.ExcelWriter("outputs/sales_analysis.xlsx", engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="Cleaned Data", index=False)
    kpi_summary.to_excel(writer, sheet_name="KPIs", index=False)
    category_analysis.reset_index().to_excel(writer, sheet_name="Category Analysis", index=False)
    region_analysis.reset_index().to_excel(writer, sheet_name="Region Analysis", index=False)
    monthly_sales.rename("Total Sales").reset_index().to_excel(writer, sheet_name="Monthly Sales", index=False)
    subcategory_profit.rename("Total Profit").reset_index().to_excel(writer, sheet_name="Subcategory Profit", index=False)
    monthly_kpi.reset_index().to_excel(writer, sheet_name="Monthly KPIs", index=False)
    top_customers.reset_index().to_excel(writer, sheet_name="Top Customers", index=False)
    
    insights_df.to_excel(writer, sheet_name="Insights & Recommendations", index=False)

print("Excel report created successfully with insights!")
```
